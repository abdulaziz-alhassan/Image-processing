```python
#Task1
import cv2
import numpy as np

img = cv2.imread('cameraman.tif')
h,w = img.shape[:2]

#1.1
scale = 2.0 
resized_img = cv2.resize(img, None, fx=scale, fy=scale, interpolation=cv2.INTER_LINEAR)
cv2.imwrite('resized_image.jpg', resized_img)
cv2.imshow('Resized Image', resized_img)
cv2.waitKey()

#1.2
center = (w // 2, h // 2)
angle = 120
scale = 1.5
rotation_matrix = cv2.getRotationMatrix2D(center, angle, scale)

cos = np.abs(rotation_matrix[0, 0])
sin = np.abs(rotation_matrix[0, 1])
new_w = int((h * sin) + (w * cos))
new_h = int((h * cos) + (w * sin))

rotation_matrix[0, 2] += (new_w / 2) - center[0]
rotation_matrix[1, 2] += (new_h / 2) - center[1]

rotated_img = cv2.warpAffine(img, rotation_matrix, (new_w, new_h))
cv2.imwrite('images/output_rotated_120.jpg', rotated_img)
cv2.imshow('Rotated 120 degrees', rotated_img)
cv2.waitKey()

#1.3
shear_factor = 0.5
shear_matrix = np.array([[1, shear_factor, 0],
                          [0, 1, 0]], dtype=np.float32)

sheared_x = cv2.warpAffine(img, shear_matrix, (int(w + shear_factor * h), h))
cv2.imwrite('images/output_sheared_x.jpg', sheared_x)
cv2.imshow('Sheared X', sheared_x)
cv2.waitKey()

shear_factor_y = 0.6
shear_matrix_y = np.float32([[1, 0, 0],
                              [shear_factor_y, 1, 0]])

sheared_y = cv2.warpAffine(img, shear_matrix_y, (w, int(h + shear_factor_y * w)))
cv2.imwrite('images/output_sheared_y.jpg', sheared_y)
cv2.imshow('Sheared Y', sheared_y)
cv2.waitKey()

```

```python
#Task2 

#2.1
negative_img = 255 - img
cv2.imwrite('images/output_negative.jpg', negative_img)
cv2.imshow('Negative Image', negative_img)
cv2.waitKey()

#2.2
c = 255 / np.log(1 + np.max(img))   
log_img = c * np.log(1 + img.astype(np.float64))
log_img = np.array(log_img, dtype=np.uint8)
cv2.imwrite('images/output_log.jpg', log_img)
cv2.imshow('Log Transformed Image', log_img)
cv2.waitKey()

#2.3
gamma = 3 
c = 1.0
img_normalized = img.astype(np.float64) / 255.0
gamma_img = c * np.power(img_normalized, gamma)
gamma_img = np.array(gamma_img * 255, dtype=np.uint8)
cv2.imwrite('images/output_gamma.jpg', gamma_img)
cv2.imshow('Power-Law (Gamma) Transformed Image', gamma_img)
cv2.waitKey()
```
