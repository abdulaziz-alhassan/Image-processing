```python
#import the libraries
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
#Sampling function
def sample_image(image, factor):

 height, width = image.shape[:2]
 sampled_image = cv2.resize(image, (width // factor, height //
 factor), interpolation=cv2.INTER_NEAREST)
 return sampled_image

#Quantization function
def quantize_image(image, levels):
  
  quantized_image = np.floor(image / (256 // levels)) * (256 //levels)
  quantized_image = quantized_image.astype(np.uint8)
  return quantized_image

#Plotting function
def plot_images(original, sampled, quantized):

 plt.figure(figsize=(12, 4))
 plt.subplot(1, 3, 1)
 plt.imshow(original, cmap='gray')
 plt.title('Original Image')
 plt.axis('off')
 plt.subplot(1, 3, 2)
 plt.imshow(sampled, cmap='gray')
 plt.title('Sampled Image')
 plt.axis('off')
 plt.subplot(1, 3, 3)
 plt.imshow(quantized, cmap='gray')
 plt.title('Quantized Image')
 plt.axis('off')
 plt.show()


```

```python
#Image path and parameters
image_path='lena_gray_256.tif'
sampling_factor=5
quantization_levels=11

#Load image in grayscale
original_image = cv2.imread(image_path, cv2.IMREAD_GRAYSCALE)
if original_image is None:
 print(f"Error: Unable to load image at {image_path}")

#Sample and quantize
sampled_image = sample_image(original_image, sampling_factor)
quantized_image = quantize_image(original_image, quantization_levels)

#Plot results
plot_images(original_image, sampled_image, quantized_image)
```

```python
#Import Image from PIL
from PIL import Image

#Open the images
img1=Image.open('lena_gray_256.tif')
img2=Image.open ('cameraman.tif')

#Change the size of the images to be the same
resize=(400,400)
img1=img1.resize(resize,Image.Resampling.LANCZOS)
img2=img2.resize(resize,Image.Resampling.LANCZOS)

#Convert the images to numpy arrays
im1arr=np.asarray(img1)
im2arr=np.asarray(img2)

#Task 2.1 : subtract two images and display the result
subtraction=im1arr-im2arr
resultImage=Image.fromarray(subtraction)
plt.show()

#Task 2.2 : add one image to a constant and display the result
constant_addition=im1arr+175 
resultImage=Image.fromarray(constant_addition)
plt.show()

#Task 2.3 : set difference operation between two gray-scale images and display the result
set_difference = im1arr - im2arr
resultImage=Image.fromarray(set_difference)
resultImage.show()

#Task 2.4 : symmetric difference operation between two gray-scale images and display the result
symmetric_difference = np.abs(im1arr - im2arr)
resultImage=Image.fromarray(symmetric_difference)
resultImage.show()

#Adding two images and display the result
addition=im1arr+im2arr
resultImage=Image.fromarray(addition)
resultImage.show()
```
